## What you'll learn

- How the Partner Selection UI lets you choose a provider for each layer
- How partner API keys are saved securely

## Overview

The workshop uses AWS services by default, and several layers can be swapped for a partner. You make those choices in the **Partner Selection UI**, a small web app that you start from this notebook.

Start it now and keep it running for the whole workshop. You don't pick every provider up front: each time you reach a new layer, come back to the UI, choose the provider for that layer, and copy the code it generates into that layer's notebook. If a partner needs an API key, the UI asks for it and stores it securely in SSM Parameter Store.

## Step 1: Check your AWS access

Confirm that your AWS credentials work, and show where your choices and keys will be saved.

```python
import boto3
import os

REGION = os.environ.get("AWS_DEFAULT_REGION", "us-east-1")  # Change to your preferred region
SSM_PREFIX = "/anycompany/agentcore"
CONFIG_PREFIX = f"{SSM_PREFIX}/config"

sts = boto3.client("sts", region_name=REGION)
ident = sts.get_caller_identity()
print(f"Account:  {ident['Account']}")
print(f"Identity: {ident['Arn']}")
print(f"Region:   {REGION}")
print(f"Selections -> {CONFIG_PREFIX}/<layer>/provider  (String)")
print(f"API keys   -> {CONFIG_PREFIX}/<layer>/api_key    (SecureString)")
```
